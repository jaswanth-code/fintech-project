# Data Cleaning & Preprocessing

This notebook performs data quality checks, missing value treatment, duplicate removal, date formatting, and preparation of analysis-ready datasets.

```python
import pandas as pd

nav = pd.read_csv("../data/raw/02_nav_history.csv")
investor = pd.read_csv("../data/raw/08_investor_transactions.csv")
```

```python
print(nav.isnull().sum())
print(investor.isnull().sum())
```

```python
nav = nav.drop_duplicates()
investor = investor.drop_duplicates()
```

```python
nav["date"] = pd.to_datetime(nav["date"])
investor["transaction_date"] = pd.to_datetime(
    investor["transaction_date"]
)
```

```python
nav = nav.sort_values(
    ["amfi_code", "date"]
)

nav["daily_return"] = (
    nav.groupby("amfi_code")["nav"]
       .pct_change()
)
```
